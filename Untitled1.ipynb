```python
import requests

response=requests.get("https://countries.dev/countries?fields=name,capital,region,subregion,population,area")
data=response.json()

print(type(data))
print(len(data))
print(data[0])
```

```python
import pandas as pd

rows=[]
for c in data:
  rows.append({
       "country_name":c.get("name"),
        "capital":c.get("capital"),
        "region":c.get("region"),
        "subregion":c.get("subregion"),
        "population":c.get("population"),
        "area":c.get("area")
  })

  df=pd.DataFrame(rows)
  df.head(10)

```

```python
import sqlite3
conn=sqlite3.connect("countries.db")
df.to_sql("countries",conn,if_exists="replace",index=False)

print("Data has been saved in SQL table")
```

```python
query1="""
select country_name,population from countries order by population desc limit 10 """

pd.read_sql(query1,conn)
```

```python
query2="""
select region,sum(population) as total_population from countries group by region order by total_population desc"""

pd.read_sql(query2,conn)
```

```python
query3="""
select region,count(*) as num_countries from countries group by region order by num_countries desc"""

pd.read_sql(query3,conn)
```

```python
query4="""
select country_name,area from countries where area>1000000 order by area desc"""

pd.read_sql(query4,conn)
```

```python
import matplotlib.pyplot as plt

region_pop=pd.read_sql(query2,conn)
plt.figure(figsize=(10,6))
plt.bar(region_pop["region"],region_pop["total_population"])
plt.xlabel("Region")
plt.ylabel("Total Population")
plt.title("Population by Region")
plt.xticks(rotation=45)
plt.show()
```
